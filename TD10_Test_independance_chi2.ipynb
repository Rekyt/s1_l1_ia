## Contexte


Ce test permet de vérifier l’absence de lien statistique entre deux variables X et Y qui sont
qualitatives.
X et Y sont dits indépendants lorsqu’il n’existe aucun lien statistique entre elles, dit autrement, la connaissance de X ne permet en aucune manière de se prononcer sur Y.

L’hypothèse nulle (H0) de ce test est la suivante : les deux variables X et Y sont indépendantes.  
L'hypothèse alternative (H1) de ce test est : les deux variables X et Y ne sont pas indépendantes.

Il est donc possible, grâce au test du $\chi ^2 $ d’indépendance, de tester le lien entre chaque paire de variables.

## Rappel de cours

Le test du $\chi^2$ d'indépendance repose sur la comparaison entre les **effectifs observés** dans un tableau croisant les deux variables (le tableau de contingence) et les **effectifs théoriques** que l'on obtiendrait si les deux variables étaient réellement indépendantes.

<img src="figures/chi2_independance_tableau_contingence.png" />

Sous H0 (indépendance), l'effectif théorique $t_{ij}$ de la case à la ligne $i$ et la colonne $j$ se calcule à partir des effectifs marginaux (les totaux de ligne et de colonne) et de l'effectif total $n$ :

$$ t_{ij} = \frac{n_{i \bullet} \times n_{\bullet j}}{n} $$

avec :
- $n_{i \bullet}$ le total de la ligne $i$ (l'effectif total de la modalité $i$ de la première variable)
- $n_{\bullet j}$ le total de la colonne $j$ (l'effectif total de la modalité $j$ de la seconde variable)
- $n$ l'effectif total de l'échantillon

La statistique du test, notée $\chi^2$, mesure l'écart global entre effectifs observés $n_{ij}$ et effectifs théoriques $t_{ij}$ :

$$ \chi^2 = \sum_{i=1}^{k}\sum_{j=1}^{l}{\frac{(n_{ij} - t_{ij})^2}{t_{ij}}} $$

avec $k$ le nombre de modalités (lignes) de la première variable et $l$ le nombre de modalités (colonnes) de la seconde.

Sous H0, cette statistique suit une loi du $\chi^2$ à $(k-1)\times(l-1)$ degrés de liberté (d.d.l.). Plus les effectifs observés s'éloignent des effectifs théoriques, plus la valeur du $\chi^2$ est grande, ce qui rend l'indépendance des deux variables (H0) de moins en moins plausible.

Comme pour le test du $\chi^2$ d'ajustement (cf. TD précédent), ce test n'est valide que si les effectifs théoriques sont suffisamment grands, ce qui est vérifié dans la section suivante.
